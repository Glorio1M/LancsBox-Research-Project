This is a set of codes to visualise word frequency obtained in a corpus analysis in LancsBox. Adjectives, conjuctions, articles have been romoved. Also, only key words have been extracted for visualisation

```python
import pandas as Pand
import matplotlib.pyplot as Pt


# Input the dataset
data = {
    "Word": ["tree", "three", "free", "teeth", "English", "problem", "learners"],
    "Frequency": [16, 12, 9, 8, 7, 4, 1]
      }


df = Pand.DataFrame(data)

# visualising the data

Pt.figure()
Pt.bar(df["Word"], df["Frequency"])
Pt.xticks(rotation = 50)
Pt.xlabel("Words")
Pt.ylabel("Frequency")
Pt.title("Corpus Words Frequency Analysis")
Pt.tight_layout()
Pt.show()




```
